## ***NumPy Array Operations***

### **1. Create a 1D NumPy Array**

```python
import numpy as np
```

```python
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(temperatures_w1)
```

### **2. Inspection and Properties**

```python
print(f'Shape {temperatures_w1.shape}')
```

```python
print(f'Data Type {temperatures_w1.dtype}')
```

```python
print(f'Number of Elements {temperatures_w1.size}')
```

### **3. Array Operations**

```python
Fahrenheit = (temperatures_w1 * 9/5) + 32
print(f'Celsius to Fahrenheit {Fahrenheit}')
```

```python
print(f'Maximum Temperature {temperatures_w1.max()}')
```

```python
print(f'Minimum Temperature {temperatures_w1.min()}')
```

```python
print(f' Average Temperature {round(temperatures_w1.mean(),1)}')
```

### **4. Array Slicing and Indexing**

```python
print(f'First Three days Temperature {temperatures_w1[0:3]}')
```

```python
print(f'last Two days Temperature {temperatures_w1[-2:]}')
```

```python
print(f'Middle three days Temperature {temperatures_w1[2:5]}')
```

### **6. Inspect and Slice the 2D Array**

```python
print(f'Data Type {temperatures.dtype}')
```

```python
print(f'Total Number of Elements {temperatures.size}')
```

```python
print(f'last two days for both Weeks: \n{temperatures[ : , 5 : ]}')
```

## ***Pandas Series***

### **1. Creating Pandas Series**

```python
import pandas as pd
```

```python
marks = pd.Series([95, 92, 89, 85, 80], index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'], name='Marks')
marks
```

### **2. Indexing and Slicing**

```python
print(f'1st Rank Student Mark {marks.iloc[0]}')
```

```python
print(f'Top 3 Ranks Mark: \n{marks.loc['Rank1' : 'Rank3']}')
```

```python
print(f'3rd Rank Student Mark {marks.iloc[2]}')
```

```python
print(f'Marks are Greater than 90 \n{marks[marks > 90]}')
```

### **3. Manipulating Series**

```python
marks.loc['Rank1'] = 100
print(f'Updated Marks: \n{marks}')
```

```python
del marks['Rank5']
print(f'Updated Marks: \n{marks}')
```

```python
CGPA = marks/10
print(f'CGPA: \n{CGPA}')
```

## ***Pandas DataFrame***

### **1. Creating Pandas DataFrame**

```python
transactions = pd.DataFrame({'TransactionID' : [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
                             'ProductCategory' : ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
                             'Region' : ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'],
                             'Amount' : [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
                             })
transactions
```

### **2. Data Exploration**

```python
transactions.head()
```

```python
transactions.tail()
```

```python
print(f'Shape {transactions.shape}')
```

```python
print(f'Columns {transactions.columns}')
```

```python
print(f'Data Types: \n{transactions.dtypes}')
```

```python
print(f'ProductCategory and Amount Columns: \n\n{transactions[['ProductCategory', 'Amount']]}')
```

```python
print(f'Last 3 Columns: \n\n{transactions.iloc[ : , -3 : ]}')
```

```python
print(f'Region {transactions['Region'].unique()}')
```

```python
transactions.groupby('Region')['Amount'].mean()
```

### **3. Manipulating the DataFrame**

```python
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
transactions
```

```python
transactions['Discount'] = transactions['Amount']/10
transactions
```

```python
transactions.drop(transactions[transactions['TransactionID'] == 109].index,inplace=True)
transactions
```

```python
del transactions['Discount']
transactions
```
