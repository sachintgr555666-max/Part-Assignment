## **Sets (Operations)**

a. Create a set called my_set with the following values: ['a','e','i','o','u','a','a','i']
Analyze the output and explain the same.

```python
my_set = {'a','e','i','o','u','a','a','i'}
```

```python
my_set
```

####**.** **A Set stores only unique values.**
####**.** **Duplicate  elements ('a' and 'i') are automatically removed.**
####**.** **Therefore, the set contains only: {'a', 'e', 'i', 'o', 'u'}**
####**.** **Sets are unordered, so the elements are not stored in a fixed order.**

b. Attempt to access my_set[4]. If the code throws an error, provide an explanation.

```python
my_set[4]
```

####**.** **Set do not support indexing because they are unordered.**
####**.** **Since there is no fixed position for elements, you can't access an element using an index like my_set[4].**

## **Operators & Conditional Statements : (IF, ELIF, ELSE)**

1. Prompt the user for input. Score range should be from 0 to 10 (both inclusive).

```python
score = int(input("Enter your score (0 to 10): "))

if score < 0 or score > 10:
  print("Invalid score. Please enter a score between 0 and 10.")
elif score > 7:
  print("Above Average: Excellent work! Keep it up!")
elif score >= 4 or score <= 7:
  print("Average: Good effort! Keep practicing, there's room for improvement.")
elif score < 4:
  print("Below Average: Need to improve your performance. Consistent practice wil lead to better results.")
else:
  print("Invalid score")
```
